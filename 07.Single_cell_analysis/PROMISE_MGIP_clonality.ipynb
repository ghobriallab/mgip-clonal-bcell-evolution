```python
import numpy as np
import pandas as pd
import scanpy as sc
import gseapy as gp
import scanpy.external as sce
import os, glob, subprocess, itertools
import anndata as ad
import matplotlib.pyplot as plt
import seaborn as sns
import re
import warnings
warnings.filterwarnings('ignore')
from statannotations.Annotator import Annotator
from scipy.stats import mannwhitneyu
from scipy.stats import pearsonr
from scipy.stats import wilcoxon
from scipy.stats import kruskal
from statsmodels.stats.multitest import multipletests
from sklearn.utils import resample
from scipy.stats import permutation_test
from scipy.stats import bootstrap
from Levenshtein import distance as levenshtein_distance
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from scipy.stats import chi2_contingency, fisher_exact
from scipy.stats import pearsonr
from sklearn.metrics import mean_squared_error
import numpy as np
from sklearn.cluster import KMeans
from scipy.stats import gaussian_kde
from scipy.signal import find_peaks
from adjustText import adjust_text

import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from adjustText import adjust_text


heavy_chain_dict = {
    "IGHM" : "IgM",
    "IGHA" : "IgA",
    "IGHG" : "IgG"
}

light_chain_dict = {
    "IGLC" : "Lambda",
    "IGKC" : "Kappa"
}

ig_dict = {
    'IgM': 'IGHM', 'IgA': 'IGHA', 'IgG': 'IGHG', 'Kappa': 'IGKC', 'Lambda': 'IGLC', 
    'IgG_Kappa': ('IGHG', 'IGKC'), 'IgG_Lambda': ('IGHG', 'IGLC'),
    'IgA_Kappa': ('IGHA', 'IGKC'), 'IgA_Lambda': ('IGHA', 'IGLC'), 
    'IgM_Kappa': ('IGHM', 'IGKC'), 'IgM_Lambda': ('IGHM', 'IGLC')
}

demux_meta_df = pd.read_csv('/home/kimsungj/02.MGIP/96.PROMISE/260425_addtional_sequenced/20260427_MGIP_PROMISE_additional_sequenced.csv', sep=",")

demux_dict = dict(zip(demux_meta_df['final_demux_id'], demux_meta_df['PROMISE_ID']))

ms_ori_df = pd.read_csv('../data/PROMISE_ms_clone.csv', sep=",")

bcr_dir = '/home/kimsungj/02.MGIP/94.VDJtoMZratio/00.Output'

pid_list = [
    1329, ## IgG Lambda
    2115, ## IgG
    2936, ## IgG Lambda (large offset: 13.2)
    2512, ## IgM
    563, ## IgM Kappa (no match)
    1106, ## IgM
    1263, ## IgM (large offset: 55.3)
    6373 ## IgM
]

demux_meta_df['PROMISE_ID'] = demux_meta_df['PROMISE_ID'].astype(int)
demux_meta_df['sample_type_sk'] = demux_meta_df['sample_type'] + "_" + demux_meta_df['cell_fraction']

### clone colors
clone_color_dict = {
 '563_clone1': '#fc8d62', '563_clone2': '#66c2a5', '563_clone3': '#8da0cb', '563_clone4': '#5874b3', '563_clone5': '#c2cce3',
 '1106_clone1': '#8da0cb', '1106_clone2': '#5874b3', '1106_clone3': '#c2cce3', '1106_clone4': '#3c5283', '1106_clone5': '#f6f8fb',
 '1263_clone1': '#8da0cb', '1263_clone2': '#5874b3', '1263_clone3': '#c2cce3', '1263_clone4': '#3c5283', '1263_clone5': '#f6f8fb',
 '1329_clone1': '#66c2a5', '1329_clone2': '#3f9d7f', '1329_clone3': '#9dd8c5', '1329_clone4': '#296653', '1329_clone5': '#d3eee5', '1329_clone6': '#133027',
 '2115_clone1': '#fc8d62', '2115_clone2': '#66c2a5', '2115_clone3': '#3f9d7f', '2115_clone4': '#9dd8c5', '2115_clone5': '#296653', '2115_clone6': '#d3eee5', '2115_clone7': '#133027',
 '2512_clone1': '#8da0cb', '2512_clone2': '#5874b3',
 '2936_clone1': '#66c2a5', '2936_clone2': '#3f9d7f', '2936_clone3': '#8da0cb',
 '6373_clone1': '#8da0cb', '6373_clone2': '#5874b3', '6373_clone3': '#c2cce3'
}

target_clones_to_show={
    1329 : ['1329_clone6'], ## IgG Lambda
    2115 : ['2115_clone2'], ## IgG
    2936 : ['2936_clone2'], ## IgG Lambda (large offset: 13.2)
    2512 : ['2512_clone1'], ## IgM
    1106 : ['1106_clone3'], ## IgM
    6373 : ['6373_clone1'], ## IgM
}

vdj_present_libraries = [
    "sc_77_S10_R1_PB_CD19pos_GEX_5",
    "sc_77_S10_R1_PB_WBC_GEX_5",
    "sc_77_S10_R2_PB_CD19pos_GEX_5",
    "sc_77_S10_R2_PB_WBC_GEX_5",
    "sc_77_S11_R1_PB_WBC_GEX_5",
    "sc_77_S11_R2_PB_CD19pos_GEX_5",
    "sc_77_S12_R1_PB_CD19pos_GEX_5",
    "sc_77_S12_R1_PB_WBC_GEX_5",
    "sc_77_S12_R2_PB_CD19pos_GEX_5",
    "sc_77_S13_R1_PB_WBC_GEX_5",
    "sc_77_S13_R2_PB_CD19pos_GEX_5",
    "sc_77_S14_R1_PB_CD19pos_GEX_5",
    "sc_77_S14_R1_PB_WBC_GEX_5",
    "sc_77_S3_R1_PB_WBC_GEX_5",
    "sc_77_S3_R2_PB_WBC_GEX_5",
    "sc_77_S4_R1_PB_WBC_GEX_5",
    "sc_77_S5_R1_BM_WBC_GEX_5",
    "sc_77_S6_R1_BM_CD138pos_GEX_5",
    "sc_77_S6_R1_BM_WBC_GEX_5",
    "sc_77_S7_R1_PB_WBC_GEX_5",
    "sc_77_S7_R2_PB_WBC_GEX_5",
    "sc_77_S9_R1_PB_WBC_GEX_5",    
]

## filtered
adata_ori = sc.read_h5ad('../data/merged_MGIP_progressing_PROMISE_pc_b_filtered.h5ad')

for pid in pid_list:
    print(f'************** {pid} **************')
    ## MS results (longitudinal)
    ms_df = ms_ori_df[ms_ori_df['PROMISE_ID'] == pid]
    ms_df['Sample_Date'] = pd.to_datetime(ms_df['Sample_Date'], errors='coerce')
    ms_df = ms_df.sort_values('Sample_Date', ascending=True)
    
    #display(ms_df[['clone_id','Sample_Date', 'Type', 'Conc', 'm.z', ]])

    pid_meta = demux_meta_df.loc[
        demux_meta_df['PROMISE_ID'] == pid,
        ['final_demux_id', 'sample_type_sk', 'date_of_sampling', 'qc_issue']
    ].drop_duplicates()
    
    if pid_meta.empty:
        print(f"[Warning] No final_demux_id found for PROMISE_ID={pid}")
        continue

    ## Get nearest MS results on BM biopsy
    target_ms = ms_df.copy()
    target_ms['Sample_Date'] = pd.to_datetime(target_ms['Sample_Date'])
    
    pid_meta['date_of_sampling'] = pd.to_datetime(pid_meta['date_of_sampling'], errors='coerce')
    target_date = pid_meta['date_of_sampling'].max()
    date_diff = (target_ms['Sample_Date'] - target_date).abs()
    nearest_date = target_ms.loc[date_diff.idxmin(), 'Sample_Date']
    ms_results = target_ms[target_ms['Sample_Date'] == nearest_date].copy()
    
#     #### earliest one 
#     earliest_date = target_ms['Sample_Date'].min()
#     ms_results = target_ms[target_ms['Sample_Date'] == earliest_date].copy()

    total_ms = ms_results[['PROMISE_ID', 'Sample_Date', 'Type', 'm.z', 'Conc']]

    #### Total MS results
    total_ms_summary = [f"{row['Type']}, {row['m.z']}, {row['Conc']}" for _, row in total_ms.iterrows()]
    print(f"Total MS results (near sequencing date) ({len(total_ms)} peaks): {(' / '.join(total_ms_summary))}")
    #print(f"MS sample: {nearest_date}, scBCR-seq: {target_date}")
    
    ############ Extract estimated m/z ratio from scBCR
    scbcr_list = []

    print("\n------ Used scBCR-seq (qc issue) ------")

    for _, meta_row in pid_meta.iterrows():
        final_demux_id = meta_row['final_demux_id']
        sample_type = meta_row['sample_type_sk']
        qc_issue = meta_row['qc_issue']
        if 'VDJ_BCR' in final_demux_id:
            print(final_demux_id, qc_issue)
            try:
                bcr_id = final_demux_id
                sample_id = re.sub(r'_VDJ.*$', '', bcr_id)
                redcap_id = pid

                file_path = f'{bcr_dir}/{bcr_id}_mzratio.csv'

                if os.path.exists(file_path):
                    tmp_df = pd.read_csv(file_path, sep=",")

                    tmp_df['sample_id'] = sample_id
                    tmp_df['redcap_id'] = redcap_id
                    tmp_df['sample_type'] = sample_type
                    tmp_df['final_demux_id'] = final_demux_id
                    tmp_df['source_file'] = os.path.basename(file_path)

                    tmp_df['barcode_pure'] = tmp_df['barcode'].str.split("BCR_").str[-1]
                    tmp_df['gex_barcode'] = (
                        tmp_df['barcode_pure'] + '-' + tmp_df['sample_id'] + "_GEX_5"
                    )

                    scbcr_list.append(tmp_df)
            except Exception as e:
                print(f"[Warning] No file: {file_path} (PROMISE_ID={pid}, final_demux_id={final_demux_id})")
        else: # GEX
            continue

    if len(scbcr_list) == 0:
        print(f"[Warning] No mzratio files found for PROMISE_ID={pid}")
        continue

    scbcr_df = pd.concat(scbcr_list, axis=0, ignore_index=True)

    scbcr_df = scbcr_df.drop_duplicates().copy()
    
    scbcr_df = scbcr_df.rename(columns={'mz_ratio': 'est_mz_ratio'})
    
    
    ## Annotation
    annot_df_new = scbcr_df[['gex_barcode', 'clonal', 'est_mz_ratio', 'full_aa_seq_filt', 'heavy_c_gene_all_contig', 'c_gene']]

    annot_df_new.set_index('gex_barcode', inplace=True)


    ##################### MS & scBCR m/z ratio match #####################
    ## Top 10 (for fast search)
    group_cols = ['redcap_id', 'heavy_c_gene_all_contig', 'c_gene', 'full_aa_seq_filt', 'est_mz_ratio',] 
    top_counts = scbcr_df.groupby(group_cols, as_index=False).size()
    #top_10_info = top_counts.sort_values(by='size', ascending=False).head(10)
    #display(top_10_info)
    top_info = top_counts[top_counts['size'] >= 5].copy()
    top_scbcr_df = pd.merge(scbcr_df, top_info, on=group_cols, how='inner') ##top_scbcr_df = pd.merge(scbcr_df, top_10_info, on=group_cols, how='inner')
    top_scbcr_df = top_scbcr_df.sort_values(by=['redcap_id', 'size'], ascending=[True, False])

    # top_scbcr_indv_df
    top_scbcr_indv_df = top_scbcr_df.copy()
    match_info_for_viz = {} 

    for _, ms_row in ms_results.iterrows():
        ms_type = ms_row['Type']
        ms_mz = ms_row['m.z']
        ms_conc = ms_row['Conc']

        # Type Parsing
        parts = ms_type.split('_')
        h_search = ig_dict.get(parts[0])
        l_search = ig_dict.get(parts[1]) if len(parts) > 1 else None

        if h_search:
            mask = top_scbcr_indv_df['heavy_c_gene_all_contig'].str.contains(h_search, na=False)
        else:
            continue

        if l_search:
            mask = mask & top_scbcr_indv_df['c_gene'].str.contains(l_search, na=False)

        matched_candidates = top_scbcr_indv_df[mask].copy()

        if not matched_candidates.empty:
            matched_candidates['mz_diff'] = (matched_candidates['est_mz_ratio'] - ms_mz).abs()

            # 1) diff < 100 
            matched_candidates = matched_candidates[matched_candidates['mz_diff'] < 100]

            if matched_candidates.empty:
                continue

            # 2) best match
            min_diff = matched_candidates['mz_diff'].min()
            best_matches = matched_candidates[matched_candidates['mz_diff'] == min_diff].copy()

            for i, match_row in best_matches.iterrows():
                e_mz = match_row['est_mz_ratio']
                diff = match_row['mz_diff']
                
                n_match = (scbcr_df['est_mz_ratio'] == e_mz).sum()
                
                sample_type_str = ",".join(sorted(best_matches['sample_type'].dropna().astype(str).unique()))

                # 2) MS Peak  (Type, MS m/z, Conc, Diff) info include in label
                label = f"Match: {ms_type} | MS m/z: {ms_mz:.1f} | Conc: {ms_conc:.2f} | VDJ N: {n_match} | Sample_type: {sample_type_str} | Diff: {diff:.1f}"

                # multiple peaks can be matched with same est_mz --> least diff prioritized
                if e_mz not in match_info_for_viz:
                    match_info_for_viz[e_mz] = label
                else:
                    existing_diff = float(match_info_for_viz[e_mz].split('Diff: ')[-1])
                    if diff < existing_diff:
                        match_info_for_viz[e_mz] = label

            print(f"\n[MS ENTRY] Type: {ms_type} | Target m/z: {ms_mz} | Conc: {ms_conc}")
        else:
            print(f"\n[MS ENTRY] Type: {ms_type} (m/z: {ms_mz}) - No matching found.")
    
    print("\n====================== Match results ======================")
    print(match_info_for_viz)
    
    ##################### scRNA visualization #####################    
    adata_ori.obs['PROMISE_ID'] = adata_ori.obs['library'].map(demux_dict)
    
    adata = adata_ori[adata_ori.obs['PROMISE_ID'] == pid]

    adata = adata[adata.obs['library'].isin(vdj_present_libraries)]
    
    ## Add m/z ratio estimation
    adata.obs = adata.obs.join(annot_df_new, how='left')

   
    ## adata gene name change
#     gene_rename_dict = {
#         'ENSG00000289536': 'IGHJ2',
#     }

#     adata.var_names = (
#         adata.var_names.to_series()
#         .map(lambda x: gene_rename_dict.get(x, x))
#         .values
#     )
    
    #### Mask genes (IG, and MT genes)
    sex_specific_malat1_genes = ['XIST', 'RPS4Y1', 'DDX3Y', 'MALAT1']
    
    mask = (
        adata.var_names.str.startswith('IG') | 
        adata.var_names.str.startswith('MT-') | 
        adata.var_names.str.startswith('RPL') | 
        adata.var_names.str.startswith('RPS') | 
        adata.var_names.isin(sex_specific_malat1_genes)
    )
    adata = adata[:, ~mask]

    #### Total-count normalization
    sc.pp.normalize_total(adata, target_sum=1e4)
    sc.pp.log1p(adata)
      
    adata.raw = adata

    ## Filter total_count
    #min_total, max_total = 200, 6000
    #adata = adata[adata.obs.total_counts < int(max_total), :]
    #adata = adata[adata.obs.total_counts > int(min_total), :]

    #scale data, clip values exceeding standard deviation 10.
    sc.pp.scale(adata, max_value=10)
    
    # CNV / Numbat 
    try:
        numbat_df = pd.read_csv(f'su2c_tumor_cnv/PROMISE_{pid}/clone_post_2.tsv', sep="\t", index_col='cell')
        numbat_df = numbat_df[['GT_opt', 'clone_opt']]
        numbat_df['clone_opt'] = numbat_df['clone_opt'].astype(str)
        adata.obs = adata.obs.join(numbat_df, how='left')
        print('Yes CNV')
    except:
        print('No CNV')

    if match_info_for_viz:
        def get_label(x):
            return match_info_for_viz.get(x, "Others")

        adata.obs['ms_match_detailed'] = adata.obs['est_mz_ratio'].apply(get_label)

        matched_labels = [l for l in adata.obs['ms_match_detailed'].unique() if l != "Others"]
        matched_labels.sort() 
        
        ## clonotype 
        clonotype_palette = dict(zip(['clonotype'+str(i) for i in range(1, 11)], sns.color_palette("deep", 10)))
        clonotype_palette['Others'] = 'lightgrey'
        ##### from enclone
#         target_clonotypes = [f'clonotype{i}' for i in range(1, 11)]
#         adata.obs['clonotype_top10'] = adata.obs['clonotype_raw'].astype(str).where(
#             adata.obs['clonotype_raw'].isin(target_clonotypes), 
#             'Others'
#         )
            
#         colname = 'clonotype_top10'
#         label_counts = adata.obs[colname].value_counts()

#         label_map = {label: f"{label} (n={label_counts[label]:,})" for label in label_counts.index}
#         adata.obs[f'{colname}_labeled'] = adata.obs[colname].map(label_map)
        
        ########## Color MS-to-BCR matched
        palette = {}
        palette['Others'] = 'lightgrey'
        
        stats_data = []
        
        label_to_cid = {}
        
        for e_mz, info in match_info_for_viz.items():
            ms_type = info.split("Match: ")[1].split(" | ")[0]
            ms_mz_str = info.split("MS m/z: ")[1].split(" | ")[0]
            target_mz_val = round(float(ms_mz_str), 1)
            mz_diff = info.split("Diff: ")[1]
            vdj_n = info.split("VDJ N: ")[1].split(" | ")[0]
            ### estimated mz match
            mask = (adata.obs['est_mz_ratio'] == e_mz)
            group_obs = adata.obs[mask]
            rna_n = len(group_obs)

            new_label = f"{ms_type} (N = {rna_n}, MS m/z = {float(ms_mz_str):,.1f}, diff = {float(mz_diff):.1f})"

            mask = adata.obs['est_mz_ratio'] == e_mz
            adata.obs.loc[mask, 'ms_match_detailed'] = new_label

            matched_ms_row = ms_results[ms_results['m.z'].round(1) == target_mz_val].iloc[0]
            cid = f"{matched_ms_row['clone_id']}"

            palette[new_label] = clone_color_dict.get(cid, 'grey')
            label_to_cid[new_label] = cid

            stats_data.append({
                'Group Name (Isotype, m/z, diff)': new_label,
                'VDJ Clones': vdj_n,
                'scRNA Cells': rna_n
            })

    
        ### clone count
        summary_df = pd.DataFrame(stats_data)
    
        if not summary_df.empty:
            print(f"\n------ Clone Match Summary: PID {pid} ------")
            display(summary_df)
        else:
            print(f"\n[Warning] No matching groups found for PID {pid}")

        ##### manual count
        aa_counts = adata.obs['full_aa_seq_filt'].value_counts()
        top10_sequences = aa_counts.head(10).index.tolist()

        seq_to_cid = {seq: f'clonotype{i+1}' for i, seq in enumerate(top10_sequences)}

        adata.obs['clonotype_top10'] = adata.obs['full_aa_seq_filt'].map(lambda x: seq_to_cid.get(x, 'Others'))
    
        counts = adata.obs['clonotype_top10'].value_counts()
        label_map = {c: f"{c} (n={counts[c]:,})" for c in counts.index}
        adata.obs['clonotype_labeled'] = adata.obs['clonotype_top10'].map(label_map)


        for label in adata.obs['clonotype_labeled'].unique():
            base_label = label.split(" (n=")[0]
            if base_label in clonotype_palette:
                clonotype_palette[label] = clonotype_palette[base_label]
            else:
                clonotype_palette[label] = 'lightgrey'  # default fallback
                
        adata.obs['Clonotype'] = adata.obs['clonotype_labeled']
        
        ## library
        adata.obs['Sample Type'] = adata.obs['library'].str.split("_").apply(lambda x: " ".join(x[-4:-2]))
        
   
        #### Run PCA and UMAP
        res=0.2

        sc.tl.pca(adata, svd_solver='arpack')
        sc.pp.neighbors(adata, n_neighbors=10, n_pcs=40)
        sc.tl.leiden(adata, resolution=float(res)) #sc.tl.lougvain(adata)

        sc.tl.paga(adata)
        sc.pl.paga(adata, plot=False)  # remove `plot=False` if you want to see the coarse-grained graph
        sc.tl.umap(adata, init_pos='paga')

        ### Visualization

        ## sort matched earlier
        adata.obs['is_matched'] = adata.obs['ms_match_detailed'] != 'Others'
        adata = adata[adata.obs['is_matched'].argsort()].copy()
        
        ## transparency
        alphas = [0.3 if label == 'Others' else 1.0 for label in adata.obs['ms_match_detailed']]
        
        plt.rcParams.update({'font.size': 20})
        fig, ax = plt.subplots()
        sc.pl.embedding(adata, basis='X_umap', 
                        color=['ms_match_detailed'], 
                        palette=palette,
                        sort_order=False,
                        size=60,
                        edgecolor='black', linewidth=0.3,
                        alpha=alphas, ax=ax, show=False,
                        title = '',
                        )
        ax.set_xlabel("UMAP1")
        ax.set_ylabel("UMAP2")
        fig.savefig(f'../00.Figures/{pid}_MStoBCRmatch.pdf', format='pdf', bbox_inches='tight')
        plt.show()
        
        sc.pl.embedding(adata, basis='X_umap', 
                        color=['Clonotype'], 
                        palette=clonotype_palette,
                        frameon=False)
        
        ########### Targeted
        if pid in target_clones_to_show:
            target_cids = target_clones_to_show[pid]
            
            def get_targeted_label(val):
                if val == 'Others': return 'Others'
                if label_to_cid.get(val, '') in target_cids:
                    return val
                return 'Others'
                
            adata.obs['ms_match_targeted'] = adata.obs['ms_match_detailed'].apply(get_targeted_label)
            
            adata.obs['is_targeted'] = adata.obs['ms_match_targeted'] != 'Others'
            adata_target = adata[adata.obs['is_targeted'].argsort()].copy()
            
            alphas_target = [0.3 if label == 'Others' else 1.0 for label in adata_target.obs['ms_match_targeted']]
            
            fig, ax = plt.subplots(figsize=(3.5, 4))
            sc.pl.embedding(adata_target, basis='X_umap', 
                            color=['ms_match_targeted'], 
                            palette=palette,      
                            sort_order=False,
                            size=60,
                            edgecolor='black', linewidth=0.3,
                            alpha=alphas_target, ax=ax, show=False,
                            title = f'',
                            )
            ax.set_xlabel("UMAP1")
            ax.set_ylabel("UMAP2")
            fig.savefig(f'../00.Figures/{pid}_MStoBCRmatch_Targeted.pdf', format='pdf', bbox_inches='tight')
            plt.show()
        
        
        ########### Sample type and B, PC
        fig, axes = plt.subplots(3, 3, figsize=(18, 10))
        plt.rcParams.update({'font.size': 20})

        sc.pl.embedding(adata, basis='X_umap', color=['Sample Type'], 
                        ax=axes[0, 0], show=False, frameon=False,
                        size=60, edgecolor='black', linewidth=0.1)
        axes[0, 1].axis('off') 
        axes[0, 2].axis('off')

        plasma_markers = ['MZB1', 'CD38', 'XBP1']
        for i, gene in enumerate(plasma_markers):
            sc.pl.umap(adata, color=gene, ax=axes[1, i], show=False, 
                       frameon=False, size=40)

        bcell_markers = ['CD79A', 'CD79B', 'CD19']
        for i, gene in enumerate(bcell_markers):
            sc.pl.umap(adata, color=gene, ax=axes[2, i], show=False, 
                       frameon=False, size=40)

        plt.subplots_adjust(hspace=0.2, wspace=0.1)
        fig.savefig(f'../00.Figures/Supple_{pid}_SampleType_B_PC.pdf', format='pdf', bbox_inches='tight')
        plt.show()
        
        ## cell quality
        sc.pl.umap(adata, color=['n_genes_by_counts', 'total_counts', 'pct_counts_mt'])
        
```
